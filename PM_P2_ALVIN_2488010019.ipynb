Apa perbedaan supervised, unsupervised, dan reinforcement learning?

Supervised Learning: belajar dari data yang memiliki label. Contoh: memprediksi siswa lulus/tidak lulus.
Unsupervised Learning: belajar dari data tanpa label untuk menemukan pola/kelompok. Contoh: mengelompokkan siswa berdasarkan kebiasaan belajar.
Reinforcement Learning: belajar melalui reward dan punishment dari interaksi dengan lingkungan. Contoh: AI belajar bermain game dengan mendapat poin saat menang.
